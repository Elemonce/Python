## 1. Counting birds

### <u> Counting birds part 1 </u>

Every year there is a day where people are encouraged to count the birds in their garden. Joseph tries to participate every year. Below you see the data of Joseph's observations. Let's make a nice plot of the data, also include the total amount of birds spotted.

```python
# Sample data
data = {
    'year': ['2020', '2021', '2022', '2023', '2024'],
    'Pigeons': [33, 52, 26, 40, 38],
    'Sparrows': [54, 48, 66, 59, 61],
    'others': [104, 98, 88, 92, 101]
}
```

### Inspecting the data

Load and inspect the data.

```python
# Load and inspect
df = pd.DataFrame(data)

df
```

### Plotting the pigeon count

Plot the number of counted pigeons over the year, following these specifications:

* Line graph
* Size of the figure: (10,6)
* Full blue line, data point marked with a circle, linewidth 2, marker size 8
* Provide the necessary labels in order to interpret the graph and show a legend
* Provide a meaningfull title
* Show a grid

```python
#Plotting the Pigeon count
plt.figure(figsize=(10, 6))
plt.plot(df["year"], df["Pigeons"], marker="o", linewidth=2, markersize=8, color="blue")
plt.xlabel("Year")
plt.ylabel("Number of pigeons")
plt.title("Number of pigeons counted over the years")
plt.grid()

plt.show()
```

### Plotting the sparrow count

Plot the number of sparrows Joseph counted over the years, adhere to the specifications listed below:

* Line graph
* linestyle: dotted
* color of the graph: Green
* marker: a triangle pointing to the right
* the edge of the markers should be red with width 1
* Corresponding labels and titles

Ensure the plot of the Pigeon count is incorporated as well.


```python
# Plotting the sparrow count
plt.figure(figsize=(10, 6))

plt.plot(df["year"], df["Sparrows"], linestyle="dotted", color="green", marker=">", markersize=8, markeredgecolor="red", markeredgewidth=1, linewidth=2)
plt.xlabel("Year")
plt.ylabel("Number of sparrows")
plt.title("Number of sparrows counted over the years")
```

### Plotting the other birds

Add a graph line for all other birds counted. You can use a styling of your choice.

```python
# Plotting the other birds count
plt.figure(figsize=(10, 6))

plt.plot(df["year"], df["others"], linewidth=2, color="red", linestyle="dashed", marker="o", markeredgecolor="blue")

plt.xlabel("Year")
plt.ylabel("Number of other birds")

plt.title("Number of other birds counted over the years")
```

### <u> Counting birds part 2 </u>

### Plotting the total of birds counted

Add a column to the dataframe containing the total amount of birds counted that year.


```python
# Total of birds
df["total"] = df["Pigeons"] + df["Sparrows"] + df["others"]

df
```



Add a **barplot** for the total of birds counted that day on top of the already created plots of the bird species. Some specifications:
* Bar width = 0.4
* Bar color = purple
* Use corresponding labels 

Use corresponding labels 

```python
plt.figure(figsize=(10, 6))

bar_width = 0.4

plt.bar(df["year"], df["total"], bar_width, color="purple", label="Total number of birds seen")
plt.legend()

plt.xlabel("Year")
plt.ylabel("Number of birds")

plt.title("Number of birds seen over the years")

plt.show()
```

```python
# Your code
fig, ax = plt.subplots(1, 2)

ax[0].bar(df["year"], df["Pigeons"], bar_width, color = "red", label = "pigeons")
ax[0].bar(df["year"], df["Sparrows"], bar_width, bottom = df["Pigeons"], color = "blue", label = "sparrows")
ax[0].bar(df["year"], df["others"], bar_width, bottom = df["Pigeons"] + df["Sparrows"], color = "yellow", label = "others")
ax[0].grid(axis="y") # or x or nothing and then lines are both for x and y


bar_width_total = 0.6
ax[1].bar(df["year"], df["total"], bar_width_total, color="purple", label = "total number of birds")
ax[1].grid(axis="y")
# can also be done on ax instead so that it's inside a particular ax rather than outside of it
fig.legend()
# ax[1].bar(df["year"], df["total"])
```
