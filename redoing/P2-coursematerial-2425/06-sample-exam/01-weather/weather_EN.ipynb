In the file `weather_dataset.csv` you find weather data for various cities. It is up to you to clean and analyse this data. Use the provided code fields under the tasks to write down your code. 

Import the necessary packages.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

* Import the weather dataset into a DataFrame named `df_weather`
* Inspect the DataFrame by showing the first 10 rows (do not use slicing for this).


```python
#Import and inspect
df_weather = pd.read_csv("weather_dataset.csv")
df_weather.head(10)
```

* Change the dates in the DATE column to pandas datetime objects.
* Only include the rows with data from 2009


```python
#datetime objects, year 2009
df_weather["DATE"] = pd.to_datetime(df_weather["DATE"], format="%Y%m%d")
df_weather = df_weather[df_weather["DATE"] > pd.to_datetime("2008-12-31")]
df_weather = df_weather[df_weather["DATE"] < pd.to_datetime("2010-01-01")]

df_weather
```

* Remove the rows with NaN values.

```python
#NaN
df_weather = df_weather.dropna()
```

* Create 4 different DataFrames, one for each of the following cities: Basel, Oslo, Roma and Maastricht. Name them `df_Basel`, `df_Oslo`, `df_Roma` and `df_Maastricht` respectively.
* Each dataframe should include the columns with the dates and months and the columns with the data of the corresponding city. 



```python
#New dataframes
city_data = {}
cities = ["Basel", "Oslo", "Roma", "Maastricht"]
for city in cities:
    city_data[city] = [col for col in df_weather.columns if city.upper() in col or (col == "DATE" or col == "MONTH")]



# df_weather[["BASEL_cloud_cover", "BASEL_temp_min"]]
# q = pd.DataFrame(df_weather[["BASEL_cloud_cover", "BASEL_temp_min"]])

df_Basel = pd.DataFrame(df_weather[[col for col in city_data["Basel"]]])
df_Oslo = pd.DataFrame(df_weather[[col for col in city_data["Oslo"]]])
df_Roma = pd.DataFrame(df_weather[[col for col in city_data["Roma"]]])
df_Maastricht = pd.DataFrame(df_weather[[col for col in city_data["Maastricht"]]])
```

### Maastricht
Plot a figure displaying following information of the temperature measurements over time done in Maastricht:
* Temperature in °C is displayed on the y-axis
* Mean temperature in orange
* Max temperature in red

* Display a legend with appropriate labels, indicating which data is shown

```python
#Maastricht
plt.figure(figsize=(10, 6))
plt.plot(df_Maastricht["DATE"], df_Maastricht["MAASTRICHT_temp_mean"], color="orange", label="mean")
plt.plot(df_Maastricht["DATE"], df_Maastricht["MAASTRICHT_temp_max"], color="red", label="max")
plt.plot(df_Maastricht["DATE"], df_Maastricht["MAASTRICHT_temp_min"], color="blue", label="min")

# {month:02} means that if the value doesn't take up two units of space, then it is preceded by a zero
plt.xticks(ticks=[f"2009-{month:02}" for month in range(1, 13)], labels=("Jan", "Feb", "March", "April", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"))
# plt.xticks(labels=("Jan", "Feb", "March", "April", "May", "June", "July", "Aug", "Sep", "Oct", "Nov", "Dec"))

plt.legend()
plt.show()

```
